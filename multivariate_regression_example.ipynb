```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
%matplotlib inline 
np.set_printoptions(precision=4, suppress=True)
```

```python
X = np.array([[66, 68, 60, 70, 65, 61, 74, 73, 75, 67],
              [42, 53, 23, 50, 20, 19, 33, 45, 43, 22],
              [22, 23, 13, 20, 10, 19, 13, 15, 23, 12]]).T
y = np.array([160, 170, 110, 178, 155, 120, 233, 215, 235, 164])
ave_x = np.average(X,axis=0)
ave_y = np.average(y)
#plot data and mean
plt.figure(figsize=(12,6))
plt.subplot(1,3,1)
plt.scatter(X[:,0],y, s=80, c="blue")
plt.scatter(ave_x[0],ave_y, s=100, c="red")

plt.subplot(1,3,2)
plt.scatter(X[:,1],y, s=80, c="cyan")
plt.scatter(ave_x[1],ave_y, s=100, c="red")

plt.subplot(1,3,3)
plt.scatter(X[:,2],y, s=80, c="green")
plt.scatter(ave_x[2],ave_y, s=100, c="red")

plt.show()
```

```python
# Now, calculate the coefficients by linear algebra
Xones = np.ones(shape=(len(X[:,0])))
Xaug = np.c_[Xones[:],X]
print(Xaug)
```

```python
XTXinv = np.linalg.inv(Xaug.T@Xaug)
theta = np.matmul(XTXinv,(Xaug.T@y))
print(theta)
```

```python


# Replace these with your data:
# X: shape (n_samples, 3), without an intercept column
# theta: [intercept, coefficient_1, coefficient_2, coefficient_3]


y_pred = theta[0] + X @ theta[1:]

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(111, projection="3d")

points = ax.scatter(
    X[:, 0], X[:, 1], X[:, 2],
    c=y_pred,
    cmap="viridis",
    s=40,
    alpha=0.8,
)

ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
ax.set_zlabel("Feature 3")
ax.set_title("Linear regression predictions")
fig.colorbar(points, ax=ax, label="Predicted output", shrink=0.7)

plt.tight_layout()
plt.show()
```
