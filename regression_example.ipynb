```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
%matplotlib inline 
```

```python
x = np.array([66, 68, 60, 70, 65, 61, 74, 73, 75, 67])
y = np.array([160, 170, 110, 178, 155, 120, 233, 215, 235, 164])
ave_x = np.average(x)
ave_y = np.average(y)
#plot data and mean
plt.scatter(x,y, s=80, c="blue")
plt.scatter(ave_x,ave_y, s=100, c="red")
plt.show()
```

```python
#first center the data points
xc = x - ave_x
yc = y - ave_y
a = xc.dot(yc)/xc.dot(xc)
b = ave_y - a*ave_x
print(ave_x,ave_y)
print(a,b)

```

```python
# make prediction
w = a*69 + b
print(w)
#plot data and found line
h=np.linspace(np.min(x)-5,np.max(x)+5,50)
line=a*h+b
plt.scatter(x,y, s=80, c="blue")
plt.scatter(67.9,174, s=100, c="red")
plt.scatter(69,a*69+b, s=100, c="black")
plt.plot(h,line,'r')
plt.xlabel('heart rate in BPM')
plt.ylabel('calories per day')
plt.show()
```
